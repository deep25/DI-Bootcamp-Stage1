```python
import pandas as pd
import numpy as np
import scipy as sp

#Evaluation of distribution shapes through skewness and kurtosis

skewness = mobile_prices.skew()
kurtosis = mobile_prices.kurtosis()

print("--- Skewness ---")
print(skewness)
print("\n--- Kurtosis ---")
print(kurtosis)
```

```python
import pandas as pd
import numpy as np
import scipy.stats as stats

ram_price_0 = mobile_prices[mobile_prices['price_range'] == 0]['ram']
ram_price_1 = mobile_prices[mobile_prices['price_range'] == 1]['ram']
ram_price_2 = mobile_prices[mobile_prices['price_range'] == 2]['ram']
ram_price_3 = mobile_prices[mobile_prices['price_range'] == 3]['ram']

stats_anova = stats.f_oneway(ram_price_0, ram_price_1, ram_price_2, ram_price_3)
print(stats_anova)

# stats_anova.pvalue gives us the calculated p-value directly
if stats_anova.pvalue < 0.05:
  print("Reject null hypothesis")
else:
  print("Fail to reject null hypothesis")
```

# Executive Summary: Key Insights and Statistical Conclusions

Based on the statistical analysis, hypothesis testing, and visualizations performed, here are the key findings regarding the determinants of mobile phone price ranges:

### 1. The Dominant Factor: RAM (Random Access Memory)
* **Finding:** `ram` is overwhelmingly the strongest determinant of the price range, with a massive correlation of **0.92**.
* **Statistical Proof:** The ANOVA test comparing RAM across the four price ranges yielded an extremely high F-statistic (~3520) and a p-value of **0.0**, forcing us to reject the null hypothesis. RAM amounts are completely distinct and non-overlapping across price categories, as clearly seen in your pink scatter plot.

### 2. Secondary Key Determinants
* **Battery Power (`battery_power`):** Correlation of **0.20**. High-end phones require robust battery capacity to support other power-hungry hardware.
* **Screen Resolution (`px_width` and `px_height`):** Correlations of **0.17** and **0.15** respectively. Better displays (higher pixel density) are moderately indicative of premium pricing.

### 3. Unexpected & Counter-intuitive Findings
* **Camera Megapixels (`pc` & `fc`):** Primary camera (`pc`, correlation **0.03**) and front camera (`fc`, correlation **0.02**) show almost **no linear relationship** with the price range. A high-megapixel count does not automatically mean a phone belongs to a higher price tier.
* **Core Count & Connectivity (`n_cores`, `four_g`, `three_g`, `wifi`):** Processor cores and modern connectivity features show negligible correlation (< 0.02) with pricing. These features have become standard baselines across all price ranges rather than premium differentiators.
