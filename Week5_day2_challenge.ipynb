```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from sklearn.datasets import make_circles

samples = 1000
X, y = make_circles(samples,
                    noise = 0.03,
                    random_state = 42)

print('X : ', X)
print('\n')
print('y : ', y)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X[:, 0], X[:, 1], c = y, cmap = plt.cm.RdYlBu)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

# Set random seed for reproducibility
tf.random.set_seed(42)

# Creating a sequential model with one dense layer
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, activation='sigmoid', input_shape=(2,))
])

# Compile the model
model.compile(loss=tf.keras.losses.BinaryCrossentropy(),
              optimizer=tf.keras.optimizers.SGD(learning_rate=0.1),
              metrics=['accuracy'])

# Fit the model
history = model.fit(X, y, epochs=100, verbose=1)

# Evaluate the model
print("\nFinal Model Evaluation:")
loss, accuracy = model.evaluate(X, y)
print(f"Loss: {loss:.4f}, Accuracy: {accuracy:.4f}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf

# Adding another layer and training
tf.random.set_seed(42)

# Define the multi-layer model
model1 = tf.keras.Sequential([
    tf.keras.layers.Dense(4, activation='relu', input_shape=(2,)),
    tf.keras.layers.Dense(1, activation='sigmoid')

])

# Compile the model using the 'adam' optimizer
model1.compile(loss=tf.keras.losses.BinaryCrossentropy(),
               optimizer='adam',
               metrics=['accuracy'])

# Fit model1 (fixed variable name from model to model1)
fit_model = model1.fit(X, y, epochs=100, verbose=1)

print("\nFinal Model Evaluation:")
loss, accuracy = model1.evaluate(X, y)
print(f"Loss: {loss:.4f}, Accuracy: {accuracy:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_decision_boundary(model, X, y):
    """
    Plots the decision boundary created by a model predicting on X.
    """
    # Define the boundaries of the grid
    x_min, x_max = X[:, 0].min() - 0.1, X[:, 0].max() + 0.1
    y_min, y_max = X[:, 1].min() - 0.1, X[:, 1].max() + 0.1

    # Generate a grid of points
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 100),
                         np.linspace(y_min, y_max, 100))

    # Make predictions on the grid points
    grid_predictions = model.predict(np.c_[xx.ravel(), yy.ravel()])

    # Check for multi-class or binary classification output shape
    if len(grid_predictions[0]) > 1:
        # Multi-class prediction
        grid_predictions = np.argmax(grid_predictions, axis=1).reshape(xx.shape)
    else:
        # Binary classification
        grid_predictions = np.round(grid_predictions).reshape(xx.shape)

    # Plot the contour boundary
    plt.contourf(xx, yy, grid_predictions, cmap=plt.cm.RdYlBu, alpha=0.7)
    plt.scatter(X[:, 0], X[:, 1], c=y, cmap=plt.cm.RdYlBu, edgecolors='k')
    plt.xlim(xx.min(), xx.max())
    plt.ylim(yy.min(), yy.max())

# Plot decision boundaries side-by-side to compare results
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.title("Model 1 (Single Dense Layer)")
plot_decision_boundary(model, X, y)

plt.subplot(1, 2, 2)
plt.title("Model 2 (Two Dense Layers)")
plot_decision_boundary(model1, X, y)

plt.show()
```

1. The Complexity of the Problem
The Data: We started with a dataset containing two circles—an inner circle (blue) and an outer circle (red).
The Challenge: A simple straight line cannot separate these two circles. To classify them correctly, the model must learn to draw a round boundary.
2. Model 1 (Single Layer): Too Simple
What we did: We built a model with just one neuron and no hidden layers.
What happened: A single neuron can only draw a straight line.
The Result: It cut the graph straight down the middle, leading to an accuracy of about 47% to 50%. This is no better than flipping a coin.
3. Model 2 (Multiple Layers with ReLU): Much Better!
What we did: We added a hidden layer with 4 neurons using the ReLU activation function.
What happened: ReLU introduces "bends" or curves. Now, the model can piece together multiple straight lines to create a complex shape.
The Result: The decision boundary became a polygon wrapping around the inner circle, and accuracy on the test set jumped to 66.5%!
4. Training vs. Testing (Generalization)
When we visualized the decision boundaries side-by-side, we saw that the boundary looked nearly identical for both the Training data and Test data.
This is great news! It means our model is generalizing well—it didn't just memorize the training set, but is successfully applying what it learned to new, unseen test data.

### 🧠 Types of Classification in Machine Learning

When training models to categorize data, we generally classify tasks into three main categories depending on how many classes there are and how labels are assigned to each sample:

---

#### 1. ⚪⚫ Binary Classification
* **Description**: The simplest form of classification where the target variable has exactly **two possible outcomes** (usually mutually exclusive).
* **Example**:
  * **Email Spam Detection**: Predicting if an email is either `Spam` (1) or `Not Spam` (0).
  * **Medical Diagnosis**: Deciding if a patient has a specific disease (`Positive` or `Negative`).

---

#### 2. 🟢🔵🟡 Multi-class Classification
* **Description**: The target variable has **three or more unique classes**, but each data point can only belong to **exactly one** class at a time.
* **Example**:
  * **Handwritten Digit Recognition (MNIST)**: Identifying a single digit from a picture as one of 10 possible classes: `0, 1, 2, 3, 4, 5, 6, 7, 8,` or `9`.
  * **Animal Classifier**: Predicting whether an image shows a `Cat`, a `Dog`, or a `Bird` (assuming there is only one animal in the photo).

---

#### 3. 🏷️🏷️ Multi-label Classification
* **Description**: The target variable has multiple unique classes, and each data point can be assigned **multiple labels simultaneously** (they are not mutually exclusive).
* **Example**:
  * **Image Tagging**: An image of a park might be labeled with several tags at once: `[Trees, Sunshine, People, Grass]`.
  * **Movie Genre Classification**: A single movie like *The Matrix* can belong to multiple genres: `[Sci-Fi, Action, Thriller]`.
