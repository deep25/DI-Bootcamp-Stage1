```python
import numpy as np

matrix_3x3 = np.arange(1, 10).reshape(3, 3)
print("Original 3x3 Matrix:")
display(matrix_3x3)

print("\nTranspose:")
display(matrix_3x3.T)

# Sum of elements along columns (axis 0)
print("\nSum of elements along columns:")
display(matrix_3x3.sum(axis=0))

# Element-wise squaring
print("\nElement-wise square:")
display(np.square(matrix_3x3))
```

```python
import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_sample = np.random.choice(productivity_before, size=10, replace=False)
print("Productivity scores of the same employees before the training program:")
print(productivity_sample)
# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)
```

```python
import numpy as np

differences = productivity_after - productivity_before

# 2. Use basic NumPy statistical functions
mean_difference = np.mean(differences)
std_difference = np.std(differences, ddof=1) # ddof=1 for sample standard deviation
sample_size = len(differences)


standard_error = std_difference / np.sqrt(sample_size)

t_stat_numpy = mean_difference / standard_error

# Print the step-by-step NumPy calculations
print("--- Manual Hypothesis Testing with NumPy ---")
print(f"Sample Size (N): {sample_size}")
print(f"Sample Mean Difference: {mean_difference:.4f}")
print(f"Sample Standard Deviation (SD): {std_difference:.4f}")
print(f"Standard Error (SE): {standard_error:.4f}")
print(f"Calculated t-statistic (NumPy): {t_stat_numpy:.4f}")


critical_t_value = 1.699

print(f"\nCritical t-value threshold (for 95% confidence): {critical_t_value}")
if t_stat_numpy > critical_t_value:
    print(f"Decision: Reject the Null Hypothesis! Since {t_stat_numpy:.2f} > {critical_t_value}, the productivity increase is statistically significant.")
else:
    print(f"Decision: Fail to Reject the Null Hypothesis. Since {t_stat_numpy:.2f} <= {critical_t_value}, we cannot confidently say the training program worked.")
```

```python
import numpy as np
import pandas as pd

# Create daily dates from Jan 1st to Dec 1st
timeseries_data = np.arange('2023-01-01', '2023-12-01', dtype='datetime64[D]')

dt_index = pd.DatetimeIndex(timeseries_data)

# Slice by actual calendar quarters (months)
slicing_1st_Q = timeseries_data[dt_index.month.isin([1, 2, 3])]
slicing_2nd_Q = timeseries_data[dt_index.month.isin([4, 5, 6])]
slicing_3rd_Q = timeseries_data[dt_index.month.isin([7, 8, 9])]
slicing_4th_Q = timeseries_data[dt_index.month.isin([10, 11, 12])]

print("--- Slicing by Monthly Quarters ---")
print(f"Q1 (January - March): {slicing_1st_Q[0]} to {slicing_1st_Q[-1]} (Total days: {len(slicing_1st_Q)})")
print(f"Q2 (April - June): {slicing_2nd_Q[0]} to {slicing_2nd_Q[-1]} (Total days: {len(slicing_2nd_Q)})")
print(f"Q3 (July - September): {slicing_3rd_Q[0]} to {slicing_3rd_Q[-1]} (Total days: {len(slicing_3rd_Q)})")
print(f"Q4 (October - November): {slicing_4th_Q[0]} to {slicing_4th_Q[-1]} (Total days: {len(slicing_4th_Q)})")
```
