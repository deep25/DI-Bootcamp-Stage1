```python
import numpy as np
import pandas as pd

np.random.seed(42)

synthetic_temps = np.random.uniform(low=-5.0, high=35.0, size=(10, 12))

cities = [f'City_{i+1}' for i in range(10)]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

df = pd.DataFrame(data=synthetic_temps, index=cities, columns=months)

annual_avg_temp = df.mean(axis=1)

# Display the rounded average temperatures for each city
annual_avg_temp.round(2)

highest_avg_temp = annual_avg_temp.idxmax()
lowest_avg_temp = annual_avg_temp.idxmin()
print(annual_avg_temp.round(2))

print(f"City with the highest average temperature: {highest_avg_temp}")
print(f"City with the lowest average temperature: {lowest_avg_temp}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

np.random.seed(42)

synthetic_temps = np.random.uniform(low=-5.0, high=35.0, size=(10, 12))

cities = [f'City_{i+1}' for i in range(10)]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

df = pd.DataFrame(data=synthetic_temps, index=cities, columns=months)

# Create dropdown widget containing individual cities plus an 'All' option
widget_city = widgets.Dropdown(
    options=['All'] + cities,
    value='All',
    description='Select City:',
)

# Create an output widget to hold the plot dynamically
plot_output = widgets.Output()

# Define the plot updating function
def update_plot(selected_city):
    with plot_output:
        clear_output(wait=True)
        plt.figure(figsize=(12, 6))

        if selected_city == 'All':
            # Plot all cities on the same graph
            for city in cities:
                plt.plot(months, df.loc[city], marker='o', alpha=0.7, label=city)
            plt.title('Monthly Temperature Trends for All Cities', fontsize=14)
            plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
        else:
            # Plot the selected single city
            plt.plot(months, df.loc[selected_city], marker='o', color='magenta', linestyle='-', linewidth=2)
            plt.title(f'Monthly Temperature Trend for {selected_city}', fontsize=14)

        plt.xlabel('Months', fontsize=12)
        plt.ylabel('Temperature (°C)', fontsize=12)
        plt.ylim(-10, 40)
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.tight_layout()
        plt.show()

# Set up the event listener
widget_city.observe(lambda change: update_plot(change['new']), names='value')

# Display both the dropdown and the interactive plot container
display(widget_city)
display(plot_output)

# Run once initially to display 'All' cities
update_plot(widget_city.value)
```
