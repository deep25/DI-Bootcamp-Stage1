```python
from scipy import stats
import numpy as np

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_statistic, p_value = stats.ttest_ind(data1, data2)

print("t-statistics:", t_statistic)
print("p-value:", p_value)
```

```python
import numpy as np

house_sizes = np.array([50, 70, 80, 100, 120])  # (in square meters)
house_prices = np.array([150000, 200000, 210000, 250000, 280000])  # (in currency units)

```

```python
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

print("Slope:", slope)
print("Intercept:", intercept)
print("R-value:", r_value)
print("P-value:", p_value)
```

```python
house_size1 = 90
house_prize1 = slope * house_size1 + intercept
print("Predicted price for a 90-square-meter house:", house_prize1)
```

```python
from scipy import stats

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_value, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)
print("F-value:", f_value)
print("P-value:", p_value)
```

Based on the p-value (2.75e-05), which is less than 0.05, we reject the null hypothesis. There is a significant difference in plant growth between the fertilizers.

if the value is greater than 0.05, we fail to reject the null hypothesis. there is no difference in plant growth between the fertilizer

```python
from scipy.stats import pearsonr

```

```python
data = ({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})
corr, _ = pearsonr(data['age'], data['income'])
print(f'Pearson Correlation Coefficient: {corr:.2f}')
```

```python
data = ({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})
corr, _ = spearmanr(data['age'], data['income'])
print(f'Spearman Rank Correlation: {corr:.2f}')
```

Question 7 - A new drug works for 70% of patients. If 10 patients take it, what is the probability that exactly 7 of them recover?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
patient = 10
drug_work = 0.7

recovery = np.arange(patient + 1)
probability = binom.pmf(recovery, patient, drug_work)
```

```python
for i, prob in zip(recovery, probability):
  print(f"{i} patients: {prob:.3f}")
```

```python
plt.bar(recovery, probability)
plt.show()
```
