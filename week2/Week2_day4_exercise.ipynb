```python
m_health = pd.read_csv('Student Mental health.csv')
anxiety_gender = m_health.groupby('Choose your gender')['Do you have Anxiety?'].value_counts(normalize=True).unstack()

plt.figure(figsize=(8, 5))
sns.barplot(x=anxiety_gender.index, y=anxiety_gender['Yes'], palette='muted', hue=anxiety_gender.index, legend=False)

plt.xticks(rotation=0)
plt.xlabel('Gender')
plt.ylabel('Proportion of Anxiety')
plt.title('Proportion of Anxiety by Gender')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
m_health['Panic Attack (Numeric)'] = m_health['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

plt.figure(figsize=(8, 5))
sns.scatterplot(x='Age', y='Panic Attack (Numeric)', data=m_health, alpha=0.6, color='purple', s=100)

plt.xlabel('Age')
plt.ylabel('Panic Attack (0 = No, 1 = Yes)')
plt.title('Scatter Plot of Panic Attack by Age')
plt.tight_layout()
plt.show()
```
